```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point, MultiPoint
import matplotlib.pyplot as plt


df=pd.read_csv("..\Signal and KPLER port mapped to freight regions for MIT") 

# 1. Filter your data
df_filtered = df[(df['PORT_NAME_DATA_GROUP'] == 'Your_Source') & 
                 (df['REGION_MAP_TYPE'] == 'XOM_FREIGHT_SUB_REGION')].copy()

# 2. Create Geometry (Ensure Lat/Lon are numeric)
gdf = gpd.GeoDataFrame(
    df_filtered, 
    geometry=gpd.points_from_xy(df_filtered.PORT_LONGITUDE_DATA_SRC, df_filtered.PORT_LATITUDE_DATA_SRC),
    crs="EPSG:4326" # Standard WGS84 coordinates
)
```
